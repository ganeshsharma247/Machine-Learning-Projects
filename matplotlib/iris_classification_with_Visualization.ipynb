```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# 1. Load the Iris dataset into a DataFrame
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))
```

```python
# 2. Data Preview and Basic Info
print("First 5 Records:\n", df.head())
print("\nDescriptive Statistics:\n", df.describe())
print("\nClass Distribution:\n", df['species'].value_counts())
```

```python
# 3. Class Distribution Bar Chart
plt.figure(figsize=(6, 4)) #This creates a new blank canvas for the plot.
#figsize=(6, 4) sets the width to 6 inches and height to 4 inches.

sns.countplot(x='species', data=df)



plt.title('Class Distribution of Iris Species')
#Adds a title to the chart.


plt.xlabel('Species')
plt.ylabel('Count')
#These set the X-axis label to "Species" and the Y-axis label to "Count".


plt.show()
```

```python
# 4. Histograms for Each Feature



df.hist(figsize=(10, 8), bins=15) #This creates histograms for all numeric columns in your DataFrame df.



plt.suptitle('Feature Distributions')
plt.show()


```

```python
# 5. Pair Plot to Visualize Class Separation



sns.pairplot(df, hue='species', corner=True)

plt.suptitle('Pair Plot of Features by Species', y=1.02)
#y=1.02 adjusts the title's position slightly above the plot for better visibility.

plt.show()
```

```python
# 7. Scatter Plot: Petal Length vs. Petal Width by Species
plt.figure(figsize=(6, 4))
sns.scatterplot(
    x='petal length (cm)',
    y='petal width (cm)',
    hue='species',    #Colors the points based on the iris species (Setosa, Versicolor, Virginica).
    data=df,
    palette='deep'

)
plt.title('Petal Length vs. Petal Width by Species')
plt.show()
```
