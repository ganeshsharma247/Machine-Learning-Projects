Predict product sales based on advertising spend:

 - TV Ads
 - Radio Ads
 - Newspaper Ads

```python
# ─── 1) Imports ────────────────────────────────────────────────────────────────

import pandas as pd #for easy DataFrame handling

from sklearn.model_selection import train_test_split #to split data into train/test sets
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error #compute evaluation metrics
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("ashydv/advertising-dataset")

print("Path to dataset files:", path)


```

```python
df = pd.read_csv("/kaggle/input/advertising-dataset/advertising.csv")
df
```

```python
x = df[['TV', 'Radio', 'Newspaper']]
y = df[['Sales']]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5

print(f"RMSE: {rmse:.3f}")
```
