## Problem Statement – Diabetes Progression Prediction

Diabetes is a chronic disease that often leads to severe health complications, increased healthcare costs, and reduced quality of life if not managed properly. Predicting how a patient's condition will progress can help clinicians and researchers design better treatment plans, allocate resources, and identify high-risk individuals earlier.

The dataset used in this project consists of **10 baseline medical attributes** of 442 patients (such as age, BMI, blood pressure, cholesterol, and blood sugar indicators), with the target variable being a **quantitative measure of disease progression after one year**. All features are normalized to eliminate scale differences, making the dataset suitable for regression modeling.

The goal of this project is to build and evaluate **regression models** (Linear Regression, Ridge Regression, and Lasso Regression) that can predict **disease progression scores** based on patient baseline measurements. Specifically, the project aims to:

* Explore how baseline health indicators (e.g., BMI, LDL cholesterol, triglycerides, blood pressure) influence the one-year outcome
* Assess and compare model performance across standard linear regression and regularized approaches (Ridge, Lasso) to address potential multicollinearity and feature selection
* Provide interpretable insights into which features are most predictive of disease progression, offering potential medical and clinical relevance

**Success Criteria:**
Model performance will be measured using **Mean Squared Error (MSE)** and **R² Score**. A successful model should minimize prediction error (low MSE) while explaining a significant portion of variance in disease progression (high R²).


```python
# 1) Import libraries for loading data, models, preprocessing, and evaluation
from sklearn.datasets import load_diabetes                        # Diabetes regression dataset
from sklearn.linear_model import LinearRegression, Ridge, Lasso   # Linear, Ridge, and Lasso models
from sklearn.model_selection import train_test_split              # To split data into train/test
from sklearn.pipeline import make_pipeline                        # To chain transforms and models
from sklearn.preprocessing import PolynomialFeatures, StandardScaler  # Generate polynomial features & scale


# 2) Load the dataset
diabetes = load_diabetes()             # Returns an object with .data (features) and .target (labels)


# 3) Separate inputs (X) and output (y)
X = diabetes.data     # shape (442, 10): ten baseline predictors (age, BMI, blood pressure, etc.)
y = diabetes.target   # shape (442,): a quantitative measure of disease progression

```

```python
# 4) Split into training and testing sets (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,     # 20% held out for final evaluation
    random_state=42    # ensures reproducibility of the split
)
```

```python
# 5) Linear Regression


lin_reg = LinearRegression()           # instantiate a plain linear regression model

lin_reg.fit(X_train, y_train)          # learn weights w that minimize squared error on training data

y_pred_lin = lin_reg.predict(X_test)   # apply the trained model to the test set

```
