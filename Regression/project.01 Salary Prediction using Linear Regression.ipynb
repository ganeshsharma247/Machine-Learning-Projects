##Project Objective

Predict a person’s salary based on years of experience



```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

```

```python
# Load dataset (upload CSV in Colab)
# https://www.kaggle.com/datasets/ravitejakotharu/salary-datacsv
# https://drive.google.com/file/d/1y8jr8jByS8bxFE737CxG1hxNlX_f69tw/view?usp=sharing


df = pd.read_csv("/home/salary_data.csv")

# View data
print(df.head())

```

```python
# Define X and y
X = df[["YearsExperience"]]
y = df["Salary"]

```

```python
# Create model
model = LinearRegression()
model.fit(X, y)

```

```python
# Prediction
pred = model.predict([[5]])
print("Predicted Salary for 5 years experience:", pred[0])
```

```python
plt.scatter(X, y, color='blue')
plt.plot(X, model.predict(X), color='red')
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")
plt.show()
```
