## Project Objective

Predict car mileage (MPG) based on engine size


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Simplified clean dataset
data = {
    "Engine_Size": [130,165,150,140,198,220,215,225,190,170],
    "MPG": [18,15,18,16,15,14,14,14,15,16]
}

df = pd.DataFrame(data)
df
```

```python
# Step 1 – Define Input & Output

X = df[["Engine_Size"]]
y = df["MPG"]
```

```python
# Step 2 – Train Model

model = LinearRegression()
model.fit(X, y)
```

```python
# Step 3 – Prediction

pred = model.predict([[200]])
print("Predicted MPG for engine size 200:", pred[0])
```

```python
# Step 4 – Visualization

plt.scatter(X, y, color='blue')
plt.plot(X, model.predict(X), color='red')
plt.xlabel("Engine Size")
plt.ylabel("MPG")
plt.title("Engine Size vs Fuel Efficiency")
plt.show()
```
